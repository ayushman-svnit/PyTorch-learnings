## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other `vstack` or side by side `hstack`
* Stacking - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimesions permuted (swapped) in a certain way.

```python
# Let's create a create a tensor
import torch
x=torch.arange(1.,10.,1)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(3,3)   # we can't write (1,7) because 1X9=9 so it can be written as 3X3 i.e. also 9
x_reshaped
```

```python
# Change the view
z = x.view(3,3)
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original)
z[:, 0] = 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
# dim = 0 : stack row-wise
# dim = 1 : stack column-wise
```

#### torch.permute() - rearranges the dimensions of a target tensors in a specified order

```python
x_original = torch.rand(size=(224,224,3))

# permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2,0,1)    # shifts axis 0->1, 1->2, 2->0
print(f"Previous shape: \n {x_original}")
print(f"New Shape: \n {x_permuted}")
```

## Reproducibility (trying to take random out of random)

In short how neural network learns:

`start with random numbers -> tensor operations -> update random numbers to try and make them better representations of the data -> repeat -> repeat -> repeat...`

To reduce the randomness in neural networks and PyTorch comes the concept of **random seed**.

Essentially, random seed 'flavour' the randomness.

```python
import torch

# create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible function
import torch

# set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

# create random tensor
random_tensor_C = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

```python
# Setup device agnostic code
device = 'cuda' if torch.cuda.is_available() else 'cpu'
device
```

```python
# Count number of GPUs
torch.cuda.device_count()
```

For PyTorch since it's capable of running compute on the GPU or CPU, it's best practice to setup device agnostic code: https://docs.pytorch.org/docs/stable//notes/cuda.html

E.g. run on GPU if available, else set to CPU

### 3. Putting tensors and models on GPU

Using a GPU results in faster calculations.

```python
# create a tensor (default on the CPU)
tensor = torch.tensor([1,2,3])

# tensor not on GPU
print(tensor, tensor.device)
```

```python
# move tensor to GPU
tensor_on_gpu = tensor.to(device)
tensor_on_gpu
```

### 4. Moving tensors back to the cpu

```python
# If tensor is on GPU, we can't transform it to NumPy
tensor_on_gpu.numpy()
```

```python
# To fix the GPU tensor with numpy issue, first set it to cpu
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
```
